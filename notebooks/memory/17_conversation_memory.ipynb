### Conversation Memory

**Definition:**
Conversation memory allows an LLM to remember previous messages and use them to maintain context in future responses.

**Workflow:**
User Message → Conversation History → LLM → Response → Update History → Next Message

**Goal:**
Maintain context across multiple conversation turns.

**Example:**
User: “My name is Arjun.”
User: “What is my name?”
AI: “Arjun.”


```python
# Conversation messages ko store garna
from langchain_core.messages import HumanMessage, AIMessage

# Local Ollama LLM use garna
from langchain_ollama import ChatOllama
```

```python
# Ollama ko local Llama 3 model load gareko
llm = ChatOllama(model="llama3:latest", temperature=0)

print("LLM loaded successfully!")
```

```python
# Current conversation ko messages store garna empty list banayeko
conversation_history = []

print(conversation_history)
```

```python
# User ko first message
user_message = "My name is Arjun."

# User ko message conversation history ma add gareko
conversation_history.append(HumanMessage(content=user_message))

print(conversation_history)
```

```python
# Conversation history LLM lai diyera response generate gareko
response = llm.invoke(conversation_history)

print(response.content)
```

```python
# Assistant ko response pani conversation history ma store gareko
conversation_history.append(AIMessage(content=response.content))

print(conversation_history)
```

```python
# User le previous message ko information use garne question sodheko
user_message = "What is my name?"

# New user message history ma add gareko
conversation_history.append(HumanMessage(content=user_message))

# Pura conversation history LLM lai diyera response generate gareko
response = llm.invoke(conversation_history)

print(response.content)
```

```python
# Second response pani conversation history ma store gareko
conversation_history.append(AIMessage(content=response.content))
```

```python
# Current conversation history herna
for message in conversation_history:

    print(f"{message.__class__.__name__}: " f"{message.content}")
```

```python
# CONVERSATION MEMORY FUNCTION
def chat_with_memory(user_message, conversation_history):

    # User ko message conversation history ma add gareko
    conversation_history.append(HumanMessage(content=user_message))

    # Pura conversation history LLM lai diyera response generate gareko
    response = llm.invoke(conversation_history)

    # Assistant ko response history ma store gareko
    conversation_history.append(AIMessage(content=response.content))

    # Generated response return gareko
    return response.content
```

```python
# Conversation history reset gareko
conversation_history = []

# First conversation
response = chat_with_memory(
    "My favorite programming language is Python.", conversation_history
)

print("Assistant:", response)
```

```python
# Previous information ko basis ma question sodheko
response = chat_with_memory(
    "What is my favorite programming language?", conversation_history
)

print("Assistant:", response)
```

```python
# Conversation history reset gareko
conversation_history = []

# Multiple conversation turns define gareko
messages = [
    "My name is Arjun.",
    "I am learning Python.",
    "I am building a RAG system.",
    "I want to run my RAG system locally.",
]

# Each message lai conversation memory sanga process gareko
for message in messages:

    response = chat_with_memory(message, conversation_history)

    print(f"\nUser: {message}")
    print(f"Assistant: {response}")
```

```python
# Previous conversation ko information recall garna question sodheko
response = chat_with_memory("What am I currently building?", conversation_history)

print("\nAssistant:", response)
```
